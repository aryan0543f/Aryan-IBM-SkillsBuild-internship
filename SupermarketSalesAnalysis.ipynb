## 1. Problem Statement
The aim of this project is to analyze supermarket sales data and find useful information about products, branches, categories, customers, payments, and ratings.

## 2. Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

# Configure settings
warnings.filterwarnings('ignore')
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette('husl')
%matplotlib inline

print("Libraries imported successfully!")
```

## 3. Load Dataset

```python
# Load the dataset
df = pd.read_csv('supermarket_sales.csv')

# Display basic information
print("Dataset loaded successfully!")
print(f"\nDataset Shape: {df.shape}")
print(f"Total Records: {df.shape[0]}")
print(f"Total Columns: {df.shape[1]}")
```

## 4. Data Exploration

```python
# Display first few rows
print("First 5 rows of the dataset:\n")
df.head()
```

```python
# Display column information
print("Dataset Information:\n")
df.info()
```

```python
# Display statistical summary
print("Statistical Summary:\n")
df.describe()
```

## 5. Data Cleaning and Preprocessing

```python
# Check for missing values
print("Missing Values:\n")
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0] if missing_values.sum() > 0 else "No missing values found!")
```

```python
# Check for duplicate rows
duplicates = df.duplicated().sum()
print(f"\nDuplicate Rows: {duplicates}")

if duplicates > 0:
    df = df.drop_duplicates()
    print(f"Removed {duplicates} duplicate rows")
```

```python
# Calculate Sales if not present
if 'Sales' not in df.columns:
    df['Sales'] = df['Quantity'] * df['Unit_Price']
    print("Sales column created: Sales = Quantity × Unit Price")
else:
    print("Sales column already exists")

# Display updated columns
print(f"\nTotal columns: {df.columns.tolist()}")
```

## 6. Data Analysis

### 6.1 Which product generates the highest sales?

```python
# Product-wise sales analysis
product_sales = df.groupby('Product')['Sales'].sum().sort_values(ascending=False)

print("Top 10 Products by Sales:\n")
print(product_sales.head(10))

print(f"\n✓ Highest selling product: {product_sales.index[0]}")
print(f"✓ Total sales: ₹{product_sales.iloc[0]:,.2f}")
```

```python
# Visualize top products
plt.figure(figsize=(12, 6))
product_sales.head(10).plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Top 10 Products by Sales', fontsize=16, fontweight='bold')
plt.xlabel('Product', fontsize=12)
plt.ylabel('Sales (₹)', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.grid(axis='y', alpha=0.3)
plt.show()
```

### 6.2 Which branch performs best?

```python
# Branch-wise sales analysis
branch_sales = df.groupby('Branch')['Sales'].sum().sort_values(ascending=False)

print("Branch-wise Sales Performance:\n")
print(branch_sales)

print(f"\n✓ Best performing branch: {branch_sales.index[0]}")
print(f"✓ Total sales: ₹{branch_sales.iloc[0]:,.2f}")

# If City column exists, show branch-city mapping
if 'City' in df.columns:
    branch_city = df.groupby(['Branch', 'City'])['Sales'].sum().sort_values(ascending=False)
    print("\nBranch-City Performance:")
    print(branch_city)
```

```python
# Visualize branch performance
plt.figure(figsize=(10, 6))
branch_sales.plot(kind='bar', color=['#ff9999', '#66b3ff', '#99ff99'], edgecolor='black')
plt.title('Branch-wise Sales Performance', fontsize=16, fontweight='bold')
plt.xlabel('Branch', fontsize=12)
plt.ylabel('Sales (₹)', fontsize=12)
plt.xticks(rotation=0)
plt.tight_layout()
plt.grid(axis='y', alpha=0.3)
plt.show()
```

### 6.3 Which category sells the most?

```python
# Category-wise sales analysis
category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

print("Category-wise Sales:\n")
print(category_sales)

print(f"\n✓ Top selling category: {category_sales.index[0]}")
print(f"✓ Total sales: ₹{category_sales.iloc[0]:,.2f}")
```

```python
# Visualize category performance
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Bar chart
category_sales.plot(kind='bar', ax=axes[0], color='coral', edgecolor='black')
axes[0].set_title('Category-wise Sales', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Category', fontsize=12)
axes[0].set_ylabel('Sales (₹)', fontsize=12)
axes[0].tick_params(axis='x', rotation=45)
axes[0].grid(axis='y', alpha=0.3)

# Pie chart
category_sales.plot(kind='pie', ax=axes[1], autopct='%1.1f%%', startangle=90)
axes[1].set_title('Category Sales Distribution', fontsize=14, fontweight='bold')
axes[1].set_ylabel('')

plt.tight_layout()
plt.show()
```

### 6.4 What is the most popular payment method?

```python
# Payment method analysis
payment_count = df['Payment_Method'].value_counts()
payment_sales = df.groupby('Payment_Method')['Sales'].sum().sort_values(ascending=False)

print("Payment Method Usage:\n")
print(payment_count)

print("\nPayment Method Sales:\n")
print(payment_sales)

print(f"\n✓ Most popular payment method: {payment_count.index[0]}")
print(f"✓ Number of transactions: {payment_count.iloc[0]}")
```

```python
# Visualize payment methods
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Transaction count
payment_count.plot(kind='bar', ax=axes[0], color='lightgreen', edgecolor='black')
axes[0].set_title('Payment Method - Transaction Count', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Payment Method', fontsize=12)
axes[0].set_ylabel('Number of Transactions', fontsize=12)
axes[0].tick_params(axis='x', rotation=45)
axes[0].grid(axis='y', alpha=0.3)

# Sales amount
payment_sales.plot(kind='bar', ax=axes[1], color='lightcoral', edgecolor='black')
axes[1].set_title('Payment Method - Total Sales', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Payment Method', fontsize=12)
axes[1].set_ylabel('Sales (₹)', fontsize=12)
axes[1].tick_params(axis='x', rotation=45)
axes[1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

### 6.5 Do Members spend more than Normal customers?

```python
# Customer type analysis
customer_stats = df.groupby('Customer_Type')['Sales'].agg(['sum', 'mean', 'count'])
customer_stats.columns = ['Total Sales', 'Average Transaction', 'Transaction Count']

print("Customer Type Analysis:\n")
print(customer_stats)

member_avg = customer_stats.loc['Member', 'Average Transaction']
normal_avg = customer_stats.loc['Normal', 'Average Transaction']

print(f"\n✓ Member average transaction: ₹{member_avg:,.2f}")
print(f"✓ Normal customer average transaction: ₹{normal_avg:,.2f}")

if member_avg > normal_avg:
    print(f"\nMembers spend ₹{member_avg - normal_avg:,.2f} more per transaction on average.")
else:
    print(f"\nNormal customers spend ₹{normal_avg - member_avg:,.2f} more per transaction on average.")
```

```python
# Visualize customer type comparison
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Average transaction
customer_stats['Average Transaction'].plot(kind='bar', ax=axes[0], color=['#8e44ad', '#3498db'], edgecolor='black')
axes[0].set_title('Average Transaction by Customer Type', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Customer Type', fontsize=12)
axes[0].set_ylabel('Average Sales (₹)', fontsize=12)
axes[0].tick_params(axis='x', rotation=0)
axes[0].grid(axis='y', alpha=0.3)

# Total sales
customer_stats['Total Sales'].plot(kind='bar', ax=axes[1], color=['#8e44ad', '#3498db'], edgecolor='black')
axes[1].set_title('Total Sales by Customer Type', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Customer Type', fontsize=12)
axes[1].set_ylabel('Total Sales (₹)', fontsize=12)
axes[1].tick_params(axis='x', rotation=0)
axes[1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

### 6.6 What is the average customer rating?

```python
# Rating analysis
avg_rating = df['Rating'].mean()
rating_distribution = df['Rating'].value_counts().sort_index()

print(f"Average Customer Rating: {avg_rating:.2f} out of 5")
print(f"\nRating Distribution:\n")
print(rating_distribution)

# Rating statistics by category
if 'Category' in df.columns:
    category_rating = df.groupby('Category')['Rating'].mean().sort_values(ascending=False)
    print("\nAverage Rating by Category:\n")
    print(category_rating)
```

```python
# Visualize ratings
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Rating distribution histogram
df['Rating'].hist(bins=20, ax=axes[0], color='gold', edgecolor='black', alpha=0.7)
axes[0].axvline(avg_rating, color='red', linestyle='--', linewidth=2, label=f'Average: {avg_rating:.2f}')
axes[0].set_title('Customer Rating Distribution', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Rating', fontsize=12)
axes[0].set_ylabel('Frequency', fontsize=12)
axes[0].legend()
axes[0].grid(alpha=0.3)

# Rating by category
if 'Category' in df.columns:
    category_rating.plot(kind='bar', ax=axes[1], color='orange', edgecolor='black')
    axes[1].set_title('Average Rating by Category', fontsize=14, fontweight='bold')
    axes[1].set_xlabel('Category', fontsize=12)
    axes[1].set_ylabel('Average Rating', fontsize=12)
    axes[1].tick_params(axis='x', rotation=45)
    axes[1].axhline(avg_rating, color='red', linestyle='--', linewidth=1, alpha=0.5)
    axes[1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()
```

## 8. Additional Insights

```python
# Sales trend analysis (if date column exists)
if 'Date' in df.columns:
    df['Date'] = pd.to_datetime(df['Date'])
    daily_sales = df.groupby('Date')['Sales'].sum()
    
    plt.figure(figsize=(14, 6))
    daily_sales.plot(kind='line', color='darkblue', linewidth=2, marker='o', markersize=4)
    plt.title('Daily Sales Trend', fontsize=16, fontweight='bold')
    plt.xlabel('Date', fontsize=12)
    plt.ylabel('Sales (₹)', fontsize=12)
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()
```

```python
# Correlation analysis
numeric_cols = df.select_dtypes(include=[np.number]).columns
correlation_matrix = df[numeric_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm', center=0, 
            square=True, linewidths=1, cbar_kws={"shrink": 0.8})
plt.title('Correlation Matrix', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```
